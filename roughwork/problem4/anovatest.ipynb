```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
```

=============================
### Creating Fake Data for ANOVA Test


```python
np.random.seed(12)

races = ['Asian', 'Black', 'Hispanic', 'Other','White']
```

```python
voter_race = np.random.choice(a= races, p=[0.05, 0.15, 0.25, 0.05, 0.5], size=1000)
```

```python
voter_age = stats.poisson.rvs(loc=18, mu=30, size=1000)
```

```python
voter_frame = pd.DataFrame({'races':voter_race, 'ages':voter_age})
groups = voter_frame.groupby('races').groups
```

```python
asian = voter_age[groups['Asian']]
black = voter_age[groups['Black']]
hispanic = voter_age[groups['Hispanic']]
other = voter_age[groups['Other']]
white = voter_age[groups['White']]
```

=========
### ANOVA Test

```python
stats.f_oneway(asian, black, hispanic, other, white)
```

There is no significant difference between age groups of different races
