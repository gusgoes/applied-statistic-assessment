```python
# ========================================
# 🎯 CORREÇÃO DA PROVA - SISTEMA AUTOMÁTICO
# ========================================

def corrigir_prova_automatica():
    """Corrige a prova usando as respostas armazenadas automaticamente"""
    
    # Gabarito correto
    gabarito = {1: "B", 2: "B", 3: "C", 4: "C", 5: "B", 
                6: "A", 7: "B", 8: "B", 9: "A", 10: "B"}
    
    print("🔍 CORRIGINDO PROVA AUTOMATICAMENTE...")
    print("===========================================\n")
    
    # Verificar se existem respostas
    if not respostas_usuario:
        print("❌ ERRO: Nenhuma resposta encontrada!")
        print("💡 Execute todas as questões primeiro, depois rode esta correção.\n")
        return
    
    # Calcular nota
    acertos = 0
    total_questoes = len(gabarito)
    
    print("📊 ANÁLISE DETALHADA:")
    print("-" * 50)
    
    for questao in range(1, 11):
        if questao in respostas_usuario:
            sua_resposta = respostas_usuario[questao].upper()
            resposta_correta = gabarito[questao]
            
            if sua_resposta == resposta_correta:
                print(f"✅ Questão {questao:2d}: {sua_resposta} ← CORRETA")
                acertos += 1
            else:
                print(f"❌ Questão {questao:2d}: {sua_resposta} (correto: {resposta_correta})")
        else:
            print(f"⚠️  Questão {questao:2d}: NÃO RESPONDIDA")
    
    # Resultado final
    nota_percentual = (acertos / total_questoes) * 100
    
    print("\n" + "="*50)
    print(f"📈 RESULTADO FINAL:")
    print(f"🎯 Acertos: {acertos}/{total_questoes}")
    print(f"📊 Nota: {nota_percentual:.1f}%")
    
    # Classificação de desempenho
    print(f"🏆 Classificação: ", end="")
    if acertos >= 9:
        print("EXPERT em Distribuição Normal! 🌟")
    elif acertos >= 7:
        print("BOM conhecimento dos conceitos! ⭐") 
    elif acertos >= 5:
        print("Conceitos básicos ok, revisar aplicações! �")
    elif acertos >= 3:
        print("Revisar fundamentos da distribuição normal! 🔄")
    else:
        print("Estudar o notebook explicativo novamente! 📝")
    
    print("="*50)
    return acertos, nota_percentual

def corrigir_prova_manual():
    """Permite digitar todas as respostas manualmente para correção"""
    
    print("✏️ CORREÇÃO MANUAL - Digite suas respostas")
    print("=========================================\n")
    
    # Gabarito correto  
    gabarito = {1: "B", 2: "B", 3: "C", 4: "C", 5: "B",
                6: "A", 7: "B", 8: "B", 9: "A", 10: "B"}
    
    respostas_manuais = {}
    
    # Coletar respostas
    for i in range(1, 11):
        while True:
            resposta = input(f"Questão {i}: ").upper().strip()
            if resposta in ['A', 'B', 'C', 'D']:
                respostas_manuais[i] = resposta
                break
            else:
                print("⚠️ Digite apenas A, B, C ou D")
    
    # Corrigir
    acertos = 0
    print("\n📊 ANÁLISE DETALHADA:")
    print("-" * 50)
    
    for questao in range(1, 11):
        sua_resposta = respostas_manuais[questao]
        resposta_correta = gabarito[questao]
        
        if sua_resposta == resposta_correta:
            print(f"✅ Questão {questao:2d}: {sua_resposta} ← CORRETA")
            acertos += 1
        else:
            print(f"❌ Questão {questao:2d}: {sua_resposta} (correto: {resposta_correta})")
    
    # Resultado final
    nota_percentual = (acertos / 10) * 100
    
    print("\n" + "="*50)
    print(f"📈 RESULTADO FINAL:")
    print(f"🎯 Acertos: {acertos}/10")
    print(f"📊 Nota: {nota_percentual:.1f}%")
    
    # Classificação
    if acertos >= 9:
        print("🏆 EXPERT em Distribuição Normal! 🌟")
    elif acertos >= 7:
        print("🏆 BOM conhecimento! ⭐")
    elif acertos >= 5:
        print("🏆 Conceitos básicos ok! 📚")
    else:
        print("🏆 Precisa revisar! 📝")
    
    print("="*50)
    return acertos, nota_percentual

# Executar correção automática
print("🎯 Duas opções de correção disponíveis:")
print("📱 corrigir_prova_automatica() - Usa as respostas que você já deu")  
print("✏️ corrigir_prova_manual() - Digite todas as respostas novamente")
print()
print("💡 Executando correção automática...")
print()

# Corrigir automaticamente
corrigir_prova_automatica()
```
